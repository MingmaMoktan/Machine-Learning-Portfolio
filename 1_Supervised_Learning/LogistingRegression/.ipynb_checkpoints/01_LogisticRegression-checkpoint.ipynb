## Data Preprocessing

```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('medical.csv')
data.head()
```

```python
# data["Unnamed: 32"].value_counts()
data = data.drop('Unnamed: 32', axis=1)
```

```python
# Since diagonsis is a categorical variable we will convert it to a numerical variable using one hot encoding. We will use the get_dummies function from pandas to do this.
columns = ['diagnosis']
data = pd.get_dummies(data, columns=columns, drop_first=True, dtype=int)
```

```python
# Since we also don't need id column we will drop it as well.
data = data.drop('id', axis=1)
```

## Define Target and Features

```python
y = data['diagnosis_M']
features = data.drop('diagnosis_M', axis=1)
```

## Model

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(features, y, test_size=0.2, random_state=42)
```

## Training

```python
model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
print("Predicted values:", y_pred[:10])
```

## Evaluation

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

accuracy = accuracy_score(y_test, y_pred)
confusion = confusion_matrix(y_test, y_pred)
report = classification_report(y_test, y_pred)
print("Accuracy:", accuracy)
print("Confusion Matrix:\n", confusion)
print("Classification Report:\n", report)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 7))
sns.heatmap(confusion, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Predicted Benign', 'Predicted Malignant'],
            yticklabels=['Actual Benign', 'Actual Malignant'])
```
