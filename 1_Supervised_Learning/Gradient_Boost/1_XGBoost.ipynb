```python
# Mnist data is already clean as they are just the pixels so we don't need to clean and process it.
mnist = fetch_openml('mnist_784', version=1, as_frame=True)
x = mnist.data
y = mnist.target.astype(int)
```

```python
# Train test and split
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
```

```python
import xgboost as xgb
clf = xgb.XGBClassifier(
    objective='multi:softmax',
    max_depth=5, # Maximum depth of each tree
    learning_rate=0.1,
    n_estimators=100, # No of trees
    subsample=0.5, # Instead of giving the model the whole 70000 dataset to learn it gives a sample of data for training
    colsample_bytree=0.8, # Feature sampling rate    
)
```

```python
clf.fit(X_train, y_train)
# 1. Get the probabilities for every class (0 through 9) for each test image
pred_prob = clf.predict_proba(X_test)

# 2. Get the final predicted class (the number the model is most sure about)
pred_cls = clf.predict(X_test)
pred_cls
```

```python
pred_prob
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test, pred_cls))
```
