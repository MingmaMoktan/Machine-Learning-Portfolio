## Data Preprocessing

```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('/home/davidsalome/workspace/Machine-Learning-Portfolio/1_Supervised_Learning/LogisticRegression/medical.csv')
data.head()
```

```python
# data["Unnamed: 32"].value_counts()
data = data.drop('Unnamed: 32', axis=1)
data['diagnosis'].value_counts()
```

```python
# Since diagonsis is a categorical variable we will convert it to a numerical variable using one hot encoding. We will use the get_dummies function from pandas to do this.
columns = ['diagnosis']
data = pd.get_dummies(data, columns=columns, drop_first=True, dtype=int)
```

```python
# Since we also don't need id column we will drop it as well.
# data = data.drop('id', axis=1)
data.info()
# data.duplicated().sum() # No duplicates in the dataset.
```

## Define Target and Features

```python
y = data['diagnosis_M']
features = data.drop('diagnosis_M', axis=1)
```

## Model

```python
import xgboost as sgb
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(features, y, test_size=0.2, random_state=42)
```

## Training

```python
# Make instance of the model
import xgboost as xgb
model = xgb.XGBClassifier(
    objective='binary:logistic',
    max_depth=5, # Maximum depth of each tree
    learning_rate=0.1,
    n_estimators=100, # No of trees
    subsample=0.5, # Instead of giving the model the whole 70000 dataset to learn it gives a sample of data for training
    colsample_bytree=0.8, # Feature sampling rate    
)

# Train the model
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
print("Predicted values:", y_pred[:10])
prob = model.predict_proba(X_test)[:,1]
```

## Evaluation

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

accuracy = accuracy_score(y_test, y_pred)
confusion = confusion_matrix(y_test, y_pred)
report = classification_report(y_test, y_pred)
print("Accuracy:", accuracy)
print("Confusion Matrix:\n", confusion)
print("Classification Report:\n", report)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 7))
sns.heatmap(confusion, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Predicted Benign', 'Predicted Malignant'],
            yticklabels=['Actual Benign', 'Actual Malignant'])
```

### BRIEF REPORT ON HYPERPARAMETERS AND PERFORMANCE
- In this medical diagnosis classification task, tuned key XGBoost hyperparameters 
- including max_depth=5 to capture feature interactions without overfitting, 
- learning_rate=0.1 to ensure steady convergence, and n_estimators=100 to provide 
- sufficient tree capacity. Subsample and colsample_bytree were set to 0.5 and 0.8 
- respectively to introduce regularization through random sampling.
- The resulting confusion matrix (70 True Negatives, 40 True Positives, 1 False Positive, 
- and 3 False Negatives) demonstrates exceptional performance with an accuracy of ~96.5%. 
- Notably, the model achieved very low false negatives (only 3 missed cases), which is critical in medical settings to avoid failing to diagnose positive patients, while maintaining high precision with only 1 false alarm.
