```python
import pandas as pd
import numpy as np
```

## Each neuron will have the following calculation
y = wx + b 

So that is why we are setting

x as input, 

w as weight 

and b as bias. 

And in this equation wx is the matrix multiplication so that means the the **row** of w and **column** of x should be same. 

## Input Layer

```python
np.random.seed(42) # This is to generate the same set of number from the random number each time we want
x = np.random.rand(3, 1) # When we did np.random() this means this is creating the matrix of the dimension 3x1
```

## Hidden layer

```python
# Weight for the hidden layer
w = np.random.rand(4, 3) # This is also the matrix that generates the matrix of the size 4x3 so that it matches the 3x3 for the input for matrix multiplication
# Here is the bias for hidden layer
b = np.random.rand(4, 1)
# Calculation in hidden layer and the result is o as output which will be passed to relu for hidden layer and the sigmoid in output
r = np.dot(w,x)+b
```

## Activation Function and Passing the output

```python
def relu(x):
    return np.maximum(0,x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

## Output layer

```python
o = relu(r)
print(o)
```

```python
# Weight and bias for the output layer
w1 = np.random.rand(1,4)
b1 = np.random.rand(1,1)
r2 = np.dot(w1,o)
o2 = sigmoid(r2)
o2
```
