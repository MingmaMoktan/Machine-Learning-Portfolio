## Create a dataset and then run.

```python
import pandas as pd
import numpy as np

# Creating a 10-row, 5-column dataset
data = {
    'department': ['IT', 'HR', 'IT', 'Finance', 'HR', 'IT', 'Finance', 'Marketing', 'Marketing', 'IT'],
    'age': [28, 34, 41, 29, 38, 25, 45, 31, 27, 39],
    'salary': [4500, 3800, 6200, 4100, 5000, 3900, 7100, 4200, 3600, 5800],
    'experience': [3, 7, 12, 4, 9, 2, 15, 5, 2, 10],
    'score': [85, 90, 95, 78, 88, 82, 99, 80, 75, 92]
}

# 10 custom row labels (employee names) these are like the ids for loc since there are no default indices
row_labels = ['alice', 'bob', 'charlie', 'david', 'eva', 'frank', 'grace', 'hannah', 'ian', 'jack']

# Build the DataFrame
df = pd.DataFrame(data, index=row_labels)

# Display the full dataset to inspect coordinates
df.head(100)
```

## Use Case 1: Locating Specific Data Points (Single Cells or Rows)
- iloc uses numeric coordinates (0-indexed).

- loc uses exact label names.

```python
print("--- USE CASE 1: LOCATING SPECIFIC POINTS ---")

# 1. Using iloc: Get the 3rd row (index 2, which is 'charlie')
print("\n1. iloc [Row 2]:")
print(df.iloc[2])
```

```python
# 2. Using loc: Get the row labeled 'charlie' directly
print("\n2. loc ['charlie']:")
print(df.loc['charlie'])
```

```python
# 3. Using iloc: Get a single specific value -> Row index 2, Column index 2 ('salary')
print("\n3. iloc [Row 2, Col 2]:")
print(df.iloc[2, 2])
```

```python
# 4. Using loc: Get a single specific value using row name and column name
print("\n4. loc ['charlie', 'salary']:")
print(df.loc['charlie', 'salary'])
```

## Use Case 2: Slicing (Ranges of Rows and Columns)
iloc Slicing: The stop index is exclusive (stops right before the number).

loc Slicing: The stop label is inclusive (includes the final row/column name!).

```python
print("\n--- USE CASE 2: SLICING RANGES ---")

# 1. Using iloc: Rows 1 through 4 (indices 1, 2, 3), and columns 0 through 2
# Notice that row 4 and column 3 are EXCLUDED because iloc stop bounds are exclusive.
print("\n1. iloc slice (rows 1:4, cols 0:3):")
print(df.iloc[1:4, 0:3])
```

```python
# 2. Using loc: Rows from 'bob' to 'eva', and columns 'department' through 'salary'
# Notice that 'eva' and 'salary' ARE INCLUDED because loc stop bounds are inclusive.
print("\n2. loc slice (row 'bob' to 'eva', col 'department' to 'salary'):")
print(df.loc['bob':'eva', 'department':'salary'])
```

Key Takeaways to Remember in your Notebook:

- df.iloc[row_num, col_num] $\rightarrow$ Always numbers.

- df.loc[row_name, col_name] $\rightarrow$ Always labels/names.

- iloc range stops before the end number.

- loc range includes the end label.
